## 1. Contains Duplicate

Return `True` if any number appears more than once. Otherwise, return `False`.

**Pattern hint:** Keep track of values you have already seen.

```python
def contains_duplicate(numbers):
    
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] == numbers[j]:
                return True
    return False

assert contains_duplicate([1, 2, 3, 1]) is True
assert contains_duplicate([1, 2, 3, 4]) is False
assert contains_duplicate([]) is False

print("Problem 1 passed")
```

## 2. Valid Anagram

Two strings are anagrams when they contain the same letters the same number of times. Return whether `first` and `second` are anagrams.

**Pattern hint:** Count how many times each character appears.

```python
def is_anagram(first, second):
    dictionary_first = {}
    dictionary_second = {}

    for i in first:
        if i in dictionary_first:
            dictionary_first[i] += 1
        else:
            dictionary_first[i] = 1
    
    for i in second:
        if i in dictionary_second:
            dictionary_second[i] += 1
        else:
            dictionary_second[i] = 1

    return dictionary_first == dictionary_second
    
assert is_anagram("listen", "silent") is True
assert is_anagram("rat", "car") is False
assert is_anagram("aab", "abb") is False

print("Problem 2 passed")
```

## 3. Two Sum: Return Indices

Return the positions of two different numbers that add up to `target`. Assume exactly one pair exists.

Example: `[2, 7, 11, 15]` with target `9` returns `(0, 1)`.

**Pattern hint:** Start with nested loops. Remember that positions and values are different.

```python
def two_sum_indices(numbers, target):

    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                return (i, j)

assert two_sum_indices([2, 7, 11, 15], 9) == (0, 1)
assert two_sum_indices([3, 2, 4], 6) == (1, 2)
assert two_sum_indices([3, 3], 6) == (0, 1)

print("Problem 3 passed")
```

## 4. Valid Palindrome

Return whether a string reads the same forward and backward. Ignore spaces, punctuation, and capitalization.

Example: `"A man, a plan, a canal: Panama"` is a palindrome.

**Pattern hint:** Use one position at the beginning and another at the end.

```python
def is_palindrome(text):
    final_word = ''

    for i in text:
        if i.isalnum() == True:
            final_word += i.lower()
    
    return final_word == final_word[::-1]

assert is_palindrome("A man, a plan, a canal: Panama") is True
assert is_palindrome("race a car") is False
assert is_palindrome("") is True

print("Problem 4 passed")
```

## 5. Best Time to Buy and Sell Stock

Each number is a stock price on a different day. Choose one day to buy and a later day to sell. Return the largest possible profit, or `0` if no profit is possible.

Example: `[7, 1, 5, 3, 6, 4]` returns `5` because you can buy at `1` and sell at `6`.

```python
def max_profit(prices):
    stack = [0]

    for i in range(len(prices)):
        for j in range(i + 1, len(prices)):
            stack.append(prices[j] - prices[i])
            
    if max(stack) < 0:
        return 0
    else:
        return max(stack)
            
assert max_profit([7, 1, 5, 3, 6, 4]) == 5
assert max_profit([7, 6, 4, 3, 1]) == 0
assert max_profit([2, 4]) == 2

print("Problem 5 passed")
```

## 6. Move Zeroes

Move every `0` to the end of the list while keeping the nonzero numbers in their original order. Change the existing list instead of creating the final answer as a separate list.

```python
def move_zeroes(numbers):

    for i in range(len(numbers)):
        if numbers[i] == 0:
            numbers.append(numbers.pop(numbers.index(numbers[i])))
    return numbers

assert move_zeroes([0, 0, 1]) == [1, 0, 0]
assert move_zeroes([0, 1, 0, 2, 0]) == [1, 2, 0, 0, 0]
assert move_zeroes([0, 0, 0]) == [0, 0, 0]
assert move_zeroes([]) == []

print("Problem 6 passed")
```

## 7. Merge Two Sorted Lists

Given two lists that are already sorted, return one new sorted list containing every value.

Example: `[1, 3, 5]` and `[2, 4, 6]` return `[1, 2, 3, 4, 5, 6]`.

**Pattern hint:** Keep one position for each list and compare the two current values.

```python
def merge_sorted_lists(left, right):
    left_pointer = 0
    right_pointer = 0
    new_list = []

    while left_pointer < len(left) and right_pointer < len(right):
        if left[left_pointer] < right[right_pointer]:
            new_list.append(left[left_pointer])
            left_pointer += 1
        else:
            new_list.append(right[right_pointer])
            right_pointer += 1

    while right_pointer < len(right):
        new_list.append(right[right_pointer])
        right_pointer += 1

    while left_pointer < len(left):
        new_list.append(left[left_pointer])
        left_pointer += 1

    return new_list


assert merge_sorted_lists([1, 3, 5], [2, 4, 6]) == [1, 2, 3, 4, 5, 6]
assert merge_sorted_lists([1, 3, 5], [2, 4, 6]) == [1, 2, 3, 4, 5, 6]
assert merge_sorted_lists([], [1, 2]) == [1, 2]
assert merge_sorted_lists([1, 2], []) == [1, 2]

print("Problem 7 passed")
```

## 8. Valid Parentheses

A string contains only `()`, `[]`, and `{}`. Return whether every opening bracket is closed by the correct bracket in the correct order.

**Pattern hint:** A stack remembers the most recent opening bracket.

```python
def valid_parentheses(text):
    stack = []

    for i in text:
        if i == '(' or i == '[' or i == '{':
            stack.append(i)

        elif i == '}':
            if len(stack) == 0 or stack[-1] != '{':
                return False
            stack.pop()

        elif i == ']':
            if len(stack) == 0 or stack[-1] != '[':
                return False
            stack.pop()

        elif i == ')':
            if len(stack) == 0 or stack[-1] != '(':
                return False
            stack.pop()

    return len(stack) == 0

assert valid_parentheses("(") is False
assert valid_parentheses("([{}])") is True
assert valid_parentheses("(]") is False
assert valid_parentheses("([)]") is False

print("Problem 8 passed")
```

## 9. Binary Search

Given a sorted list, return the position of `target`. Return `-1` when the target is not present.

Try to solve it without checking every value one by one.

**Pattern hint:** Examine the middle value and discard half of the remaining search area.

```python
def binary_search(numbers, target):
    left = 0
    right = len(numbers) - 1

    while left <= right:
        middle = (left + right) // 2
        if numbers[middle] == target:
            return middle
            
        elif numbers[middle] < target:
            left = middle + 1

        else:
            right = middle - 1

    return -1 
        
assert binary_search([-1, 0, 3, 5, 9, 12], 9) == 4
assert binary_search([-1, 0, 3, 5, 9, 12], 2) == -1
assert binary_search([5], 5) == 0

print("Problem 9 passed")
```
