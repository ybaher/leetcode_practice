## Problem 1: Count Positive Numbers

Return how many numbers in the list are greater than `0`.

Example: `[-2, 5, 0, 3]` should return `2`.

```python
def count_positive(numbers):
    total = 0
    for i in numbers:
        if i > 0:
            total += 1
    return total

assert count_positive([-2, 5, 0, 3]) == 2
assert count_positive([-1, -5, 0]) == 0

print("Problem 1 passed")
```

## Problem 2: Find the Biggest Number

Return the biggest number in a list.

Example: `[4, 9, 2]` should return `9`.

```python
def find_biggest(numbers):
    biggest = numbers[0]

    for i in numbers:
        if i > biggest:
            biggest = i
    return biggest

assert find_biggest([4, 9, 2]) == 9
assert find_biggest([-10, -3, -7]) == -3

print("Problem 2 passed")
```

## Problem 3: Get Even Numbers

Return a new list containing only the even numbers.

Example: `[1, 2, 3, 4]` should return `[2, 4]`.

```python
def get_even_numbers(numbers):
    even_numbers = []
    for i in numbers:
        if i % 2 == 0:
            even_numbers.append(i)            
    return even_numbers

assert get_even_numbers([1, 2, 3, 4]) == [2, 4]
assert get_even_numbers([7, 9, 11]) == []

print("Problem 3 passed")
```

## Problem 4: Reverse a String

Return the string backwards.

Example: `"hello"` should return `"olleh"`.

```python
def reverse_string(text):
    text = text[::-1]
    return text

assert reverse_string("hello") == "olleh"
assert reverse_string("a") == "a"

print("Problem 4 passed")
```

## Problem 5: First Duplicate

Return the first item that appears more than once.

Return `None` if there are no duplicates.

Example: `["a", "b", "a"]` should return `"a"`.

```python
def first_duplicate(items):
    seen = []

    for i in items:
        if i in seen:
            return i
        else:
            seen.append(i)

assert first_duplicate(["a", "b", "a"]) == "a"
assert first_duplicate([1, 2, 3]) is None

print("Problem 5 passed")
```

## Problem 6: Double Every Number

Return a new list where every number is doubled.

Example: `[1, 2, 3]` should return `[2, 4, 6]`.

```python
def double_numbers(numbers):
    doubled = []
    for i in numbers:
        i *= 2
        doubled.append(i)
    return doubled


assert double_numbers([1, 2, 3]) == [2, 4, 6]
assert double_numbers([-1, 0, 5]) == [-2, 0, 10]

print("Problem 6 passed")
```

## Problem 7: Remove Vowels

Return a new string with all vowels removed.

Example: `"leetcode"` should return `"ltcd"`.

```python
def remove_vowels(text):
    vowels = ['A', 'E', 'I', 'O', 'U', 'a', 'e', 'i', 'o', 'u']
    for i in text:
        if i in vowels:
            text = text.replace(i, '')
    return text

assert remove_vowels("leetcode") == "ltcd"
assert remove_vowels("Algorithm") == "lgrthm"

print("Problem 7 passed")
```

## Problem 8: Two Sum, Easy Version

Return two numbers from the list that add up to the target.

Return `None` if no pair exists.

Example: `numbers = [2, 7, 11, 15]`, `target = 9` should return `(2, 7)`.

```python
def two_sum_easy(numbers, target):
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                return (numbers[i], numbers[j])
            else:
                return None

assert two_sum_easy([2, 7, 11, 15], 9) == (2, 7)
assert two_sum_easy([1, 2, 3], 10) is None


print("Problem 8 passed")
```
