```python
def longestCommonPrefix(strs):
    prefex = strs[0]
    for i in strs:
        while not i.startswith(prefex):
            prefex = prefex[:-1]
            if prefex == "":
                return ""
    return prefex

strs = ["flower","flow","flight"]
longestCommonPrefix(strs)
```

```python
def isValid(s):
    stack = []
    opening = ["{", "(", "["]

    pairs = {
        ")": "(",
        "}": "{",
        "]": "["
    }

    for i in s:
        if i in opening:
            stack.append(i)
        else:
            if len(stack) == 0:
                return False

            if stack[-1] == pairs[i]:
                stack.pop()
            else:
                return False

    return len(stack) == 0
```

```python
def mergeTwoLists(list1, list2):
        """
        :type list1: Optional[ListNode]
        :type list2: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        new_list = []
        while len(list1) > 0 and len(list2) > 0:
            if list1[0] < list2[0]:
                  new_list.append(list1[0])
                  list1.remove(list1[0])
            else:
                  new_list.append(list2[0])
                  list2.remove(list2[0])
        return new_list

list1 = [1,2,4]
list2 = [1,3,4]
mergeTwoLists(list1, list2)
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution(object):
    def mergeTwoLists(self, list1, list2):
        dummy = ListNode()
        current = dummy
        while list1 is not None and list2 is not None:
            if list1.val < list2.val:
                  current.next = list1
                  current = current.next
                  list1 = list1.next
            else:
                  current.next = list2
                  current = current.next
                  list2 = list2.next

        if list1 is not None:
             current.next = list1
        if list2 is not None:
             current.next = list2
        return dummy.next

list1 = [1,2,4]
list2 = [1,3,4]
mergeTwoLists(list1, list2)
```
